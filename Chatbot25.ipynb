```python
import random

greetings = ["hi", "hello", "hey", "hola", "greetings", "what's up"]
goodbyes = ["bye", "goodbye", "see you", "adios", "later"]
thanks = ["thank you", "thanks", "thx", "appreciate it"]

responses = {
    "greeting": ["Hello!", "Hi there!", "Hey! How can I help?", "Greetings!"],
    "goodbye": ["Goodbye!", "See you later!", "Bye! Have a great day!"],
    "thanks": ["You're welcome!", "No problem!", "Happy to help!", "Anytime!"],
    "default": [
        "I'm not sure I understand.",
        "Could you rephrase that?",
        "Interesting! Tell me more.",
        "Hmm... I need to think about that."
    ]
}

def chatbot_response(user_input):
    user_input = user_input.lower().strip()

    if any(greet in user_input for greet in greetings):
        return random.choice(responses["greeting"])

    elif any(goodbye in user_input for goodbye in goodbyes):
        return random.choice(responses["goodbye"])

    elif any(thank in user_input for thank in thanks):
        return random.choice(responses["thanks"])


    else:
        return random.choice(responses["default"])

print("🤖 ChatBot: Hi! I'm your iitb chatbot. Type 'quit' to exit.")
while True:
    user_input = input("You: ")
    if user_input.lower() in ["quit", "exit", "bye"]:
        print("🤖 ChatBot:", random.choice(responses["goodbye"]))
        break
    response = chatbot_response(user_input)
    print("🤖 ChatBot:", response)
```
