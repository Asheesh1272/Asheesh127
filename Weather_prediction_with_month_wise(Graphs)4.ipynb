```python
import zipfile
import pandas as pd
import io
import os

zip_file_path = "/content/archive (5).zip"
# These filenames are inferred from the original traceback's file paths.
# You might need to verify the exact filenames inside your 'archive (5).zip'
# by using `zipfile.ZipFile(zip_file_path, 'r').namelist()`
main_data_csv_name = "weather_prediction_dataset.csv"
bbq_labels_csv_name = "weather_prediction_bbq_labels.csv"

df = None
df_bbq = None

try:
    with zipfile.ZipFile(zip_file_path, 'r') as zf:
        file_list = zf.namelist()
        print(f"Files inside {zip_file_path}: {file_list}")

        # Read the main dataset
        if main_data_csv_name in file_list:
            with zf.open(main_data_csv_name) as f:
                df = pd.read_csv(io.TextIOWrapper(f, 'utf-8'))
            print(f"Successfully loaded '{main_data_csv_name}' into 'df'.")
        else:
            print(f"Warning: '{main_data_csv_name}' not found in '{zip_file_path}'. 'df' remains empty or previous value.")

        # Read the BBQ labels dataset
        if bbq_labels_csv_name in file_list:
            with zf.open(bbq_labels_csv_name) as f:
                df_bbq = pd.read_csv(io.TextIOWrapper(f, 'utf-8'))
            print(f"Successfully loaded '{bbq_labels_csv_name}' into 'df_bbq'.")
        else:
            print(f"Warning: '{bbq_labels_csv_name}' not found in '{zip_file_path}'. 'df_bbq' remains empty or previous value.")

except FileNotFoundError:
    print(f"Error: The zip file '{zip_file_path}' was not found. Please check the path.")
except KeyError as e:
    print(f"Error: A file expected inside the zip was not found: {e}. Please ensure the CSV names are correct.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

```python
for label in cols[:-1]:
    plt.hist(df[df['MONTH']==1][label],color='blue',label='January',alpha=0.1,density=True)
    plt.hist(df[df['MONTH']==2][label],color='red',label='February',alpha=0.2,density=True)
    plt.hist(df[df['MONTH']==3][label],color='green',label='March',alpha=0.3,density=True)
    plt.hist(df[df['MONTH']==4][label],color='orange',label='April',alpha=0.4,density=True)
    plt.hist(df[df['MONTH']==5][label],color='black',label='May',alpha=0.5,density=True)
    plt.hist(df[df['MONTH']==6][label],color='cyan',label='June',alpha=0.6,density=True)
    plt.hist(df[df['MONTH']==7][label],color='yellow',label='July',alpha=0.7,density=True)
    plt.hist(df[df['MONTH']==8][label],color='magenta',label='August',alpha=0.8,density=True)
    plt.hist(df[df['MONTH']==9][label],color='pink',label='September',alpha=0.9,density=True)
    plt.hist(df[df['MONTH']==10][label],color='olive',label='October',alpha=0.45,density=True)
    plt.hist(df[df['MONTH']==11][label],color='grey',label='November',alpha=0.75,density=True)
    plt.hist(df[df['MONTH']==12][label],color='purple',label='December',alpha=0.16,density=True)
    plt.ylabel("Month")
    plt.xlabel(label)
    plt.legend()
    plt.show()
```
