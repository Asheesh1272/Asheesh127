```python
import random
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
TONE = [
    "I'm here to help! 😊",
    "Let me think about that...",
    "That's a great question!",
    "Hmm... interesting!",
    "I'll do my best to answer that.",
    "Thanks for asking — I love helping out!"
]

responses = {
    "greeting": [
        "Hello! I'm MiniChatGPT — your friendly AI assistant. How can I help?",
        "Hi there! Ready to chat, learn, or just vent? I’m all ears 👂",
        "Hey! Ask me anything — from math to memes, I’ve got you."
    ],
    "goodbye": [
        "Goodbye! Come back anytime — I’ll be here with coffee and answers ☕",
        "See you soon! Don’t forget to smile today 😊",
        "Bye! Remember: You’re smarter than you think."
    ],
    "thanks": [
        "You're welcome! Let me know if you need more help.",
        "Happy to assist! Go crush your day 💪",
        "No problem! That’s what I’m here for."
    ],
    "math": [
        "Let me crunch those numbers for you!",
        "Math? Easy peasy — unless it’s calculus. Then… good luck 😉",
        "Numbers don’t lie… but sometimes they confuse me too."
    ],
    "joke": [
        "Why don’t scientists trust atoms? They make up everything!",
        "I told my computer I needed a break… now it won’t stop sending me vacation ads.",
        "What do you call a fake noodle? An impasta! 🍝"
    ],
    "motivation": [
        "You’ve got this! One step at a time.",
        "Remember: Even small progress is still progress.",
        "Don’t give up — you’re closer than you think!"
    ],
    "default": [
        "I’m not sure I understand. Could you rephrase that?",
        "Interesting! Tell me more — I’m curious.",
        "Hmm… let me think. Maybe try asking something else?"
    ]
}


last_user_input = ""

def parse_math(user_input):

    match = re.search(r'(-?\d+\.?\d*)\s*([\+\-\*/])\s*(-?\d+\.?\d*)', user_input)
    if match:
        num1 = float(match.group(1))
        op = match.group(2)
        num2 = float(match.group(3))
        try:
            if op == '+': return num1 + num2
            elif op == '-': return num1 - num2
            elif op == '*': return num1 * num2
            elif op == '/':
                if num2 == 0: return "Division by zero? Not even AI can handle that 😅"
                return num1 / num2
        except: pass
    return None

def detect_intent(user_input):
    user_input = user_input.lower()
    if any(word in user_input for word in ["hi", "hello", "hey", "greet"]):
        return "greeting"
    elif any(word in user_input for word in ["bye", "goodbye", "quit", "exit"]):
        return "goodbye"
    elif any(word in user_input for word in ["thank", "thanks", "thx"]):
        return "thanks"
    elif any(word in user_input for word in ["solve", "calculate", "what is", "plus", "minus", "times", "divided by"]):
        return "math"
    elif any(word in user_input for word in ["joke", "funny", "laugh", "humor"]):
        return "joke"
    elif any(word in user_input for word in ["motivate", "inspire", "encourage", "help me"]):
        return "motivation"
    else:
        return "default"

def generate_response(user_input):
    global last_user_input
    last_user_input = user_input

    intent = detect_intent(user_input)

    if intent == "greeting":
        return random.choice(responses["greeting"]) + " " + random.choice(TONE)
    elif intent == "goodbye":
        return random.choice(responses["goodbye"])
    elif intent == "thanks":
        return random.choice(responses["thanks"])
    elif intent == "math":
        result = parse_math(user_input)
        if result is not None:
            return f"Answer: {result} 🧮\n{random.choice(responses['math'])}"
        else:
            return "I can solve basic math — try something like 'what is 5 + 3' or 'calculate 10 * 2'."
    elif intent == "joke":
        return random.choice(responses["joke"]) + " 😄"
    elif intent == "motivation":
        return random.choice(responses["motivation"]) + " 💪"
    else:
        return random.choice(responses["default"])


print("🤖 MiniChatGPT: Hello! I’m your lightweight AI assistant.")
print("Type 'quit' to exit. Ask me anything — math, jokes, motivation, or just chat!")
print("-" * 60)

while True:
    user_input = input("You: ").strip()
    if user_input.lower() in ["quit", "exit", "bye"]:
        print("🤖 MiniChatGPT:", random.choice(responses["goodbye"]))
        break
    response = generate_response(user_input)
    print("🤖 MiniChatGPT:", response)
```
